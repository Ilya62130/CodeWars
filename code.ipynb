Дано число n. Возьмите сумму цифр числа n. Если полученное значение состоит более чем из одной цифры, продолжайте таким же образом уменьшать его, пока не получится однозначное число. Входные данные — неотрицательное целое число.

````python
def digital_root(n:int) -> int:
    number = abs(n)
    if 0 <= number < 10:
        return number
    else:
        number = str(number)
        number = sum([int(d) for d in number])
        return digital_root(number)
````

Дан массив (arr) в качестве аргумента. Завершите функцию `countSmileys`, которая должна возвращать общее количество улыбающихся лиц.

Правила для улыбающегося лица:

- Каждое улыбающееся лицо должно содержать допустимую пару глаз. Глаза могут быть обозначены как `:` или `;`.
- У улыбающегося лица может быть нос, но это не обязательно. Допустимые символы для носа: `-` или `~`.
- У каждого улыбающегося лица должен быть улыбающийся рот, обозначенный либо `)`, либо `D`.
- Никакие дополнительные символы не допускаются, кроме упомянутых.

Примеры допустимых улыбающихся лиц: `:)` `:D` `;-D` `:~)`
Недопустимые улыбающиеся лица: `;(` `:>` `:}` `:]`

Пример:
```javascript
countSmileys([':)', ';(', ';}', ':-D']);       // должно вернуть 2;
countSmileys([';D', ':-(', ':-)', ';~)']);     // должно вернуть 3;
countSmileys([';]', ':[', ';*', ':$', ';-D']); // должно вернуть 1;

В случае пустого массива верните 0. Вас не будут тестировать с некорректным вводом (ввод всегда будет массивом). Порядок элементов лица (глаза, нос, рот) всегда будет одинаковым.

````python
#Моя реализация задания, которая прошла проверку на сайте CodeWars

def count_smileys(arr:list[str]) -> int:
    res = 0
    access_charter_eys = [':', ';']
    access_charter_nose = ['-', '~']
    access_charter_smile = [')', 'D']
    for smile in arr:
        lenght = len(smile)
        if not 2 <= lenght <= 3:
            continue 

        if smile[0] not in access_charter_eys:
            continue
        
        if lenght == 2:
            if smile[1] not in access_charter_smile:
                continue
        if lenght == 3:
            if smile[1] not in access_charter_nose:
                continue
            if smile[2] not in access_charter_smile:
                continue
        res += 1        
            
    return res
````

Напишите функцию, которая принимает на вход строку и возвращает первый символ, который нигде в строке не повторяется.

Например, если на вход подана строка "stress", функция должна вернуть 't', так как буква t встречается в строке только один раз и встречается первой в строке.

В качестве дополнительного усложнения: символы в верхнем и нижнем регистре считаются одним и тем же символом, но функция должна возвращать правильный регистр для исходного символа. Например, для входа "sTreSS" функция должна вернуть "T".

Если строка содержит только повторяющиеся символы, верните пустую строку ("");

Примечание: несмотря на название в некоторых языках, ваша функция должна обрабатывать любые кодовые точки Unicode:

text
"@#@@*"    --> "#"
"かか何"   --> "何"
"🐐🦊🐐" --> "🦊"

````python
def first_non_repeating(string):
    counts = {}
    for ch in string:
        key = ch.casefold()          # приводим к одному регистру для подсчёта
        counts[key] = counts.get(key, 0) + 1

    for ch in string:
        if counts[ch.casefold()] == 1:
            return ch                # возвращаем символ в исходном регистре

    return ""
````

Чтобы при шифровке или дешифровке новый индекс остался в границах длины алфавита — получите нужный индекс через вычисление остатка при делении на 33 (при помощи оператора %).

````python
class CipherMaster:
    # Не изменяйте и не перемещайте эту переменную
    alphabet = 'абвгдеёжзийклмнопрстуфхцчшщъыьэюя'

    def cipher(self, original_text, shift):
        # Метод должен возвращать зашифрованный текст
        # с учетом переданного смещения shift.
        result = []
        for letter in original_text:
            index_char_alphabet = self.alphabet.find(letter.lower())
            if index_char_alphabet == -1:
                result.append(letter)
            else:
                new_index = (index_char_alphabet + shift) % 33
                result.append(self.alphabet[new_index])
        return ''.join(result)

    def decipher(self, cipher_text, shift):
        # Метод должен возвращать исходный текст
        # с учётом переданного смещения shift.
        result = []
        for letter in cipher_text:
            index_char_alphabet = self.alphabet.find(letter.lower())
            if index_char_alphabet == -1:
                result.append(letter)
            else:
                new_index = (index_char_alphabet - shift) % 33
                result.append(self.alphabet[new_index])
        return ''.join(result)


cipher_master = CipherMaster()
print(cipher_master.cipher(
    original_text='Однажды ревьюер принял проект с первого раза, с тех пор я его боюсь',
    shift=2
))
print(cipher_master.decipher(
    cipher_text='Олебэи яфвнэ мроплж сэжи — э пэй рдв злййвкпш лп нвящывнэ',
    shift=-3
))
````
