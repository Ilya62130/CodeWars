Цифровой корень — это рекурсивная сумма всех цифр числа.

Дано число n. Возьмите сумму цифр числа n. Если полученное значение состоит более чем из одной цифры, продолжайте таким же образом уменьшать его, пока не получится однозначное число. Входные данные — неотрицательное целое число.


16      -->  1 + 6 = 7


942     -->  9 + 4 + 2 = 15  -->  1 + 5 = 6


132189  -->  1 + 3 + 2 + 1 + 8 + 9 = 24  -->  2 + 4 = 6


493193  -->  4 + 9 + 3 + 1 + 9 + 3 = 29  -->  2 + 9 = 11  -->  1 + 1 = 2



```python
def digital_root(n:int) -> int:
    number = abs(n)
    if 0 <= number < 10:
        return number
    else:
        number = str(number)
        number = sum([int(d) for d in number])
        return digital_root(number)
```

```python
#Вариант кода с CodeWars
#Почти тоже самое что и у меня просто в одну строчку
def digital_root_codewars(n:int) -> int:
    return n if n < 10 else digital_root_codewars(sum(map(int,str(n))))
```

```python
#Вариант кода Codex
#Основан на метематике, очень быстрое выполнение

def digital_root_codex(n:int) -> int:
    return 0 if n == 0 else 1 + (n - 1) % 9


digital_root_codex(12348534)
```
